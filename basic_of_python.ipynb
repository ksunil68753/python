```python
# int
x = 42
print(x.bit_length())  # 6

# float
y = 3.0
print(y.is_integer())  # True

# complex
z = 2 + 3j
print(z.conjugate())  # (2-3j)

# string
print("python".upper())  # PYTHON

# bool
print(int(True))  # 1

# list
fruits = ["apple"]
fruits.append("banana")
print(fruits)  # ['apple', 'banana']

# tuple
t = (1, 2, 2)
print(t.count(2))  # 2

# range
print(list(range(1,5)))  # [1, 2, 3, 4]

# dict
student = {"name": "Ani"}
print(student.get("name"))  # Ani

# set
colors = {"red"}
colors.add("blue")
print(colors)  # {'red', 'blue'}

# frozenset
fs = frozenset({1, 2})
print(fs.union({2, 3}))  # {1, 2, 3}

# bytes
b = b"Hello"
print(b.decode())  # Hello

# bytearray
ba = bytearray(b"Hi")
ba.append(33)
print(ba)  # bytearray(b'Hi!')

# memoryview
mv = memoryview(b"ABC")
print(mv.tolist())  # [65, 66, 67]

# NoneType
print(type(None))  # <class 'NoneType'>
```

You define a function using the `def` keyword, followed by the function name, parentheses `()`, and a colon `:`. The code block for the function is indented.

```python
# Define a function named my_function
def my_function():
  print("Hello from a function")
```

To call or execute the function, you use the function name followed by parentheses:

```python
# Call the function
my_function()
```

You can also pass data, known as arguments, into a function by placing them inside the parentheses during the function call. The function definition includes parameters to receive these arguments.

```python
# Define a function with a parameter
def greet(name):
  print(f"Hello, {name}!")

# Call the function with an argument
greet("Alice")
```

Functions can also return values using the `return` keyword.

```python
# Define a function that returns a value
def add(x, y):
  return x + y

# Call the function and store the result
result = add(5, 3)
print(result)
```

```python
#A class is like a blueprint for creating object, an object has properties and methods (functions) associated with it. Almost everything in python is an object

#create class
class user:
    #constructor
    def __init__(self, name, email, age):
        self.name = name
        self.email = email
        self.age = age

    def greeting(self):
        return f'my name is {self.name} and I am {self.age}'

    def has_birthday(self):
        self.age+= 1

    #extend class
class customer(user):
     #constructor
    def __init__(self, name, email, age):
     self.name = name
     self.email = email
     self.age = age
     self.balance = 0 # Initialize balance for customer


    def set_balance(self, balance):
        self.balance = balance

    def greeting(self):
        return f'my name is {self.name} and i am {self.age} and my balance is {self.balance}'

#init user object
brad = user('brad Traversy', 'brad@gmail.com', 35)
#init customer object
janet = customer('janer johnson', 'janet@yahoo.com', 25)

janet.set_balance(500)
print(janet.greeting())

brad.has_birthday()
print(brad.greeting())
```
